```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
import os
from pathlib import Path
sys.path.append('../results')

pkg_path = str(Path(os.path.abspath('')).parent.absolute())
sys.path.insert(0, pkg_path)
# print(pkg_path)
data_path=pkg_path+'/functional_results/'
```

```python
# Read the data without headers (header=None)
file_name = 'DTI_feat_loss.csv'
DTI_loss_log = pd.read_csv(data_path + file_name)

DTI_loss_log = DTI_loss_log.sort_values(by=['epoch', 'batch'])

DTI_loss_log['custom_batch_value'] = (DTI_loss_log['epoch'] - 1) * 1.0 + (DTI_loss_log['batch'] // 10) * 0.5 + 0.5
print(DTI_loss_log.head())
```

```python
# Read the data without headers (header=None)
file_name = 'DSC_feat_loss.csv'
DSC_loss_log = pd.read_csv(data_path + file_name)

DSC_loss_log= DSC_loss_log.sort_values(by=['epoch', 'batch'])

DSC_loss_log['custom_batch_value'] = (DSC_loss_log['epoch'] ) * 1.0 + (DSC_loss_log['batch'] // 10) * 0.5 + 0.5
print(DSC_loss_log.head())
```

```python

fig, ax = plt.subplots(1, 2, figsize=(14, 7))  

# Plot for the first subplot (DTI Loss)
ax[0].plot(DTI_loss_log['custom_batch_value'], DTI_loss_log['train_loss'], label='Train Loss', color='goldenrod')
ax[0].plot(DTI_loss_log['custom_batch_value'], DTI_loss_log['val_loss'], label='Validation Loss', color='darkmagenta')
ax[0].grid(True)
# Labeling the first plot
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Loss')
ax[0].set_title('DTI_AD_NC U-Net trained on features: train and validation Loss')
ax[0].legend()
ax[0].set_xlim(0, 1000)

# Plot for the second subplot (DSC Loss)
ax[1].plot(DSC_loss_log['custom_batch_value'], DSC_loss_log['train_loss'], label='Train Loss', color='khaki')
ax[1].plot(DSC_loss_log['custom_batch_value'], DSC_loss_log['val_loss'], label='Validation Loss', color='steelblue')

# Labeling the second plot
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Loss')
ax[1].set_title('DSC_ap_rCBV_ED U-Net trained on features: train and validation Loss')
ax[1].legend()
ax[1].set_xlim(0, 1000)

# Show the plot with grid
plt.grid(True)
plt.show()
```

```python
# Read the data without headers (header=None)
file_name = 'DTI_surv_loss.csv'
DTI_loss_surv = pd.read_csv(data_path + file_name)

DTI_loss_surv = DTI_loss_surv.sort_values(by=['epoch', 'batch'])
DTI_loss_surv['custom_batch_value'] = (DTI_loss_surv['epoch'] ) * 1.0 + (DTI_loss_surv['batch'] // 10) * 0.5 + 0.5

print(DTI_loss_surv.head())
```

```python
file_name = 'DSC_surv_loss.csv'
DSC_loss_surv = pd.read_csv(data_path + file_name)

DSC_loss_surv = DSC_loss_surv.sort_values(by=['epoch', 'batch'])
DSC_loss_surv['custom_batch_value'] = (DSC_loss_surv['epoch'] ) * 1.0 + (DSC_loss_surv['batch'] // 10) * 0.5 + 0.5

print(DSC_loss_surv.head())
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 7))

# Plot for DTI validation loss (subplot 1)
ax[0].plot(DTI_loss_surv['custom_batch_value'], DTI_loss_surv['val_loss'], label='Validation Loss', color='darkmagenta')
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Validation Loss')
ax[0].set_title('DTI Model - Validation Loss')
ax[0].legend()
ax[0].grid(True)
ax[0].set_xlim(0, 1000)  # Adjust the x-axis limit if needed

# Plot for DSC validation loss (subplot 2)
ax[1].plot(DSC_loss_surv['custom_batch_value'], DSC_loss_surv['val_loss'], label='Validation Loss', color='steelblue')
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Validation Loss')
ax[1].set_title('DSC Model - Validation Loss')
ax[1].legend()
ax[1].grid(True)
ax[1].set_xlim(0, 1000)  # Adjust the x-axis limit if needed

# Show the plot
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 7))
ax[0].plot(DTI_loss_surv['custom_batch_value'], DTI_loss_surv['train_loss'], label='Training Loss', color='goldenrod')
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Training Loss')
ax[0].set_title('DTI Model - Training Loss')
ax[0].legend()
ax[0].grid(True)
ax[0].set_xlim(0, 1000)  # Adjust the x-axis limit if needed

# Plot for DSC training loss (subplot 2)
ax[1].plot(DSC_loss_surv['custom_batch_value'], DSC_loss_surv['train_loss'], label='Training Loss', color='khaki')
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Training Loss')
ax[1].set_title('DSC Model - Training Loss')
ax[1].legend()
ax[1].grid(True)
ax[1].set_xlim(0, 1000)  # Adjust the x-axis limit if needed

# Show the plot
plt.tight_layout()
plt.show()
```
