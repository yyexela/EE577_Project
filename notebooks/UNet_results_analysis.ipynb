```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
import os
from pathlib import Path
sys.path.append('../results')

pkg_path = str(Path(os.path.abspath('')).parent.absolute())
sys.path.insert(0, pkg_path)
# print(pkg_path)
data_path=pkg_path+'/results/'
```

```python
import pandas as pd

# Read the data without headers (header=None)
file_name = 'DTI_feat_loss.csv'
DTI_loss_log = pd.read_csv(data_path + file_name)

DTI_loss_log = DTI_loss_log.sort_values(by=['epoch', 'batch'])

DTI_loss_log['custom_batch_value'] = (DTI_loss_log['epoch'] - 1) * 1.0 + (DTI_loss_log['batch'] // 10) * 0.5 + 0.5
print(DTI_loss_log.head())
```

```python

plt.figure(figsize=(10, 6))

# Plot train_loss
plt.plot(DTI_loss_log['custom_batch_value'], DTI_loss_log['train_loss'], label='Train Loss', color='goldenrod')

# Plot val_loss
plt.plot(DTI_loss_log['custom_batch_value'], DTI_loss_log['val_loss'], label='Validation Loss', color='darkmagenta')

# Labeling the plot
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('DTI_AD_NC U-Net trained on features: train and validation Loss')
plt.legend()
plt.xlim(0, 1000)

# Show the plot
plt.grid(True)
plt.show()
```
